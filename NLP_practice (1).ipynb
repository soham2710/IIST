```python
import nltk
import re
import string
import pandas as pd
from collections import Counter
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk import pos_tag, ne_chunk
from nltk.util import ngrams
from nltk.sentiment import SentimentIntensityAnalyzer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
nltk.download('vader_lexicon')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
text = """Natural Language Processing is amazing. NLTK is a powerful Python library for NLP tasks. It supports tokenization, stemming, tagging, and parsing."""
```

```python
sentences = sent_tokenize(text)
print("\nSentence Tokenization:")
print(sentences)
```

```python
len(sentences)
```

```python
words = word_tokenize(text)
print("Word Tokenization:")
print(words)
```

```python
len(words)
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    return text

cleaned_text = clean_text(text)
print("\nCleaned Text:")
print(cleaned_text)
```

```python
stop_words = set(stopwords.words('english'))
filtered_words = [w for w in word_tokenize(cleaned_text) if w not in stop_words]

print("\nAfter Stopword Removal:")
print(filtered_words) # Normalization
```

```python
porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer("english")

print("\nPorter Stemmer:")
print([porter.stem(w) for w in filtered_words])

print("\nLancaster Stemmer:")
print([lancaster.stem(w) for w in filtered_words])

print("\nSnowball Stemmer:")
print([snowball.stem(w) for w in filtered_words])
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_words]

print("\noriginal Words:")
print(cleaned_text)

print("\nLemmatization:")
print(lemmatized)
```

```python
pos_tags = pos_tag(filtered_words)
print("\nPOS Tagging:")
print(pos_tags)

# https://medium.com/@faisal-fida/the-complete-list-of-pos-tags-in-nltk-with-examples-eb0485f04321
```

```python
# N grams are continuous sequences of N words

bigrams = list(ngrams(filtered_words, 2))
trigrams = list(ngrams(filtered_words, 3))
unigrams = list(ngrams(filtered_words, 1))

print("\nUnigrams:")
print(unigrams)

print("\nBigrams:")
print(bigrams)

print("\nTrigrams:")
print(trigrams)
```

```python
# Vocabulary - Set of all unique words

vocab = set(trigrams)
print("\nVocabulary:")
print(vocab)
```

```python
freq_dist = Counter(filtered_words)
print("\nWord Frequency:")
print(freq_dist)
```

```python
### Bag of words

vectorizer = CountVectorizer()
bow = vectorizer.fit_transform([cleaned_text])

print("\nBag of Words Features:")
print(vectorizer.get_feature_names_out())
print(bow.toarray())
```

```python
# Simple sentence splitting without NLTK (to avoid download issues)
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

text = """Natural Language Processing is amazing.
NLTK is a powerful Python library for NLP tasks.
It supports tokenization, stemming, tagging, and parsing."""

# Simple split by period
sentences = [s.strip() for s in text.split('.') if s.strip()]

# Create TF-IDF embeddings
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(sentences)

embedding_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)

embedding_df
```
